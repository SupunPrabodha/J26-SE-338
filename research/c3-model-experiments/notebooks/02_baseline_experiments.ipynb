### Objective:
Establish the empirical **performance floor** by training and evaluating traditional ML pipelines:
1. **TF-IDF + Logistic Regression**
2. **TF-IDF + Calibrated Linear Support Vector Machine (SVM)**

Evaluated on the exact 20% held-out test splits for **Dreaddit** (Stress) and **GoEmotions** (Emotion).

```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Ensure src is discoverable
if Path('../src').resolve().exists():
    sys.path.append(str(Path('../src').resolve()))

from baseline_models import run_all_baselines
```

## 1. Run Baseline Training & Evaluation

```python
results = run_all_baselines(output_dir="../results/baselines")
```

## 2. Summarize Performance Floor (Macro F1 & Accuracy)

```python
summary_data = []
for dataset_name, models in results.items():
    for model_name, metrics in models.items():
        summary_data.append({
            'Dataset': dataset_name.capitalize(),
            'Model': f"TF-IDF + {model_name}",
            'Accuracy': metrics['accuracy'],
            'Macro F1': metrics['f1_macro'],
            'Weighted F1': metrics['f1_weighted'],
            'ECE (Calibration)': metrics.get('expected_calibration_error', 'N/A')
        })

df_summary = pd.DataFrame(summary_data)
df_summary
```

## 3. Visualize Baseline Macro F1 vs Accuracy

```python
fig, ax = plt.subplots(figsize=(8, 4))
df_summary.plot(x='Model', y=['Accuracy', 'Macro F1'], kind='bar', ax=ax, rot=15)
plt.title('Baseline ML Models Performance Floor')
plt.ylabel('Score')
plt.ylim(0, 1.0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
