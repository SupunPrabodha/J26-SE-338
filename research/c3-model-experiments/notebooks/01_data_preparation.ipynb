```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Add src/ to path if working in notebook
if Path('../src').resolve().exists():
    sys.path.append(str(Path('../src').resolve()))

from data_loader import load_dreaddit_dataset, load_goemotions_dataset
```

## 1. Load & Split Dreaddit Dataset (Stress Classification)

```python
dreaddit_train, dreaddit_test = load_dreaddit_dataset(data_dir="../data/dreaddit", test_size=0.20, random_state=42)

print(f"Dreaddit Train Set: {len(dreaddit_train)} rows")
print(f"Dreaddit Test Set:  {len(dreaddit_test)} rows")
print("\nClass Distribution (Train):")
print(dreaddit_train['label'].value_counts(normalize=True))

dreaddit_train.head()
```

## 2. Load & Split GoEmotions Dataset (Emotion Classification)

```python
goemotions_train, goemotions_test = load_goemotions_dataset(data_dir="../data/goemotions", mapping="ekman", test_size=0.20, random_state=42)

print(f"GoEmotions Train Set: {len(goemotions_train)} rows")
print(f"GoEmotions Test Set:  {len(goemotions_test)} rows")
print("\nClass Distribution (Train):")
print(goemotions_train['label'].value_counts())

goemotions_train.head()
```

## 3. Dataset Distribution Verification (Plotting)

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Dreaddit split balance
dreaddit_train['label'].value_counts().plot(kind='bar', ax=axes[0], color=['#4C72B0', '#DD8452'])
axes[0].set_title('Dreaddit Train Label Distribution')
axes[0].set_xticklabels(['Non-Stress (0)', 'Stress (1)'], rotation=0)
axes[0].set_ylabel('Sample Count')

# GoEmotions split balance
goemotions_train['label'].value_counts().sort_index().plot(kind='bar', ax=axes[1], color='#55A868')
axes[1].set_title('GoEmotions (Ekman) Train Label Distribution')
axes[1].set_xlabel('Emotion Category')
axes[1].set_ylabel('Sample Count')

plt.tight_layout()
plt.show()
```
